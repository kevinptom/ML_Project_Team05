```python
X = df.drop("quality", axis=1)
y = df["quality"]

print("Number of input features:", X.shape[1])
print("Target variable: quality")
```

```python
X_train_pool, X_test, y_train_pool, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training pool:", X_train_pool.shape)
print("Fixed test set:", X_test.shape)
```

```python
training_sizes = [20, 40, 60, 80, 100]

print("Training data sizes:", training_sizes)
```

```python
results = []

print("Results storage created successfully!")
```

```python
for size in training_sizes:

    # Determine number of training samples
    n_samples = int(len(X_train_pool) * size / 100)

    # Select training subset
    X_train = X_train_pool.iloc[:n_samples].copy()
    y_train = y_train_pool.iloc[:n_samples].copy()

    # --------------------------------
    # FEATURE SCALING
    # --------------------------------

    scaler = StandardScaler()

    # Fit scaler ONLY on training data
    X_train_scaled = scaler.fit_transform(X_train)

    # Apply the same scaler to the fixed test data
    X_test_scaled = scaler.transform(X_test)

    # --------------------------------
    # LINEAR REGRESSION
    # --------------------------------

    model = LinearRegression()

    model.fit(X_train_scaled, y_train)

    # --------------------------------
    # PREDICTION
    # --------------------------------

    predictions = model.predict(X_test_scaled)

    # --------------------------------
    # EVALUATION
    # --------------------------------

    mae = mean_absolute_error(y_test, predictions)

    rmse = np.sqrt(
        mean_squared_error(y_test, predictions)
    )

    r2 = r2_score(y_test, predictions)

    # --------------------------------
    # STORE RESULTS
    # --------------------------------

    results.append({
        "Training Size (%)": size,
        "Training Samples": n_samples,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

    print(f"{size}% experiment completed successfully!")

print("All 5 experiments completed successfully!")
```

```python
results_df = pd.DataFrame(results)

results_df
```

```python
results_df.to_csv(
    "training_data_results_scaled.csv",
    index=False
)

print("New experimental results saved successfully!")
```
